This is a live example showing that the same graph can be drawn in various ways. 
Instructions: proceed through this example by pressing shift-enter or the play button in the toolbar above.

First, we create a simple directed graph with five nodes and six edges.

```python
import networkx as nx
import matplotlib.pyplot as plt
from networkx.drawing.nx_pydot import graphviz_layout

# 构建图
G = nx.DiGraph()
G.add_edge("a", "b")
G.add_edge("b", "c")
G.add_edge("c", "d")
G.add_edge("d", "e")
G.add_edge("e", "c")
G.add_edge("a", "d")
```

Now, let's draw it. We first draw it using circular layout.

```python
# Circo 布局（Graphviz）
pos = graphviz_layout(G, prog="circo")

plt.figure(figsize=(6, 6))
nx.draw(
    G, pos,
    with_labels=True,
    arrows=True,
    node_size=1500,
    node_color="#A0CBE2",
    font_size=12,
    arrowstyle="-|>",
    arrowsize=20
)
plt.show()
```

Then, let's use a layered layout.

```python
pos = graphviz_layout(G, prog="dot")

plt.figure(figsize=(6, 6))
nx.draw(
    G, pos,
    with_labels=True,
    arrows=True,
    node_size=1500,
    node_color="#A0CBE2",
    font_size=12,
    arrowstyle="-|>",
    arrowsize=20
)
plt.show()
```

Finally, let's draw it using a different layout.

```python
pos = graphviz_layout(G, prog="neato")

plt.figure(figsize=(6, 6))
nx.draw(
    G, pos,
    with_labels=True,
    arrows=True,
    node_size=1500,
    node_color="#A0CBE2",
    font_size=12,
    arrowstyle="-|>",
    arrowsize=20
)
plt.show()
```
