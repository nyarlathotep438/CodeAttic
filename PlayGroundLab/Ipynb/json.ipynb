# 1.把 Python 对象转成 JSON 字符串（序列化）

```python
import json

data = {"name": "Shixiang", "age": 26}

json_str = json.dumps(data)
print(json_str)
```

# 2.把 JSON 字符串转回 Python 对象（反序列化）

```python
import json

json_str = '{"name": "Shixiang", "age": 26}'

data = json.loads(json_str)
print(data)
print(data["name"])
```

# 3. 从文件读取 JSON

```python
import json

with open("config.json", "r") as f:
    config = json.load(f)

print(config)
print(config["lr"])
```

# 4. 写入 JSON 文件

```python
import json

data = {"model": "GNN", "version": 3}

with open("output.json", "w") as f:
    json.dump(data, f, indent=4)
```
