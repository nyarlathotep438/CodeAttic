# 环境变量设置
用于notebook等无法方便传达命令行变量的场合，能够方便统一管理（使用命令行工具时我比较喜欢用yaml）。

```python
import os

os.environ["MY_FLAG"] = "1"

def check_flag():
    if os.environ.get("MY_FLAG") == "1":
        print("Flag is ON")
    else:
        print("Flag is OFF")

check_flag()
```

```python
os.environ["MY_FLAG"] = "0"
```

```python
def check_flag():
    if os.environ.get("MY_FLAG") == "1":
        print("Flag is ON")
    else:
        print("Flag is OFF")

check_flag()
```
